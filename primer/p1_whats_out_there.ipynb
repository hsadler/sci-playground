# P1 · What's out there

A tour of the contents of the universe, from the ground up. The goal isn't to memorise a
taxonomy — it's so that when you read "M dwarf", "eclipsing binary", "AGN" or "kilonova"
later, you have somewhere to put it.

Three ideas do most of the work in this notebook:

1. **Everything is organised by scale**, and the scales are absurd. Intuition built at
   human scale is useless, so we use logarithms and stop trying to picture it.
2. **Distance is time.** Light is slow compared to the universe. Looking far away *is*
   looking into the past — this is not a metaphor, it's the observing condition.
3. **Most things you can observe are either a star, made of stars, or the corpse of a
   star.** Stars are the reference object, which is why P2 is entirely about them.

## The hierarchy

Roughly smallest to largest, with what each one is:

| Object | What it is | Where you meet it |
|---|---|---|
| **Moons, asteroids, comets** | Rocky/icy bodies orbiting something bigger | Occultations, rotation light curves |
| **Planets** | Big enough to be round, not big enough to fuse | Transits — this repo |
| **Brown dwarfs** | Too big to be planets, too small to fuse hydrogen (~13–80 Jupiters) | The awkward middle of the mass function |
| **Stars** | Self-gravitating balls of plasma fusing hydrogen | Everything |
| **Binaries & multiples** | Two or more stars orbiting each other. Roughly *half* of all stars | Eclipsing binaries — the "impostors" of notebook 04 |
| **Stellar remnants** | White dwarfs, neutron stars, black holes: what's left after fusion stops | X-ray binaries, pulsars, gravitational waves |
| **Star clusters** | Hundreds to millions of stars born together | Open clusters (young), globulars (ancient) |
| **Nebulae / the ISM** | Gas and dust between stars — where stars form and where they return | Emission lines, dust extinction |
| **Galaxies** | 10⁸–10¹² stars plus gas, dust and dark matter | Everything extragalactic |
| **Active galactic nuclei** | A supermassive black hole actively feeding, outshining its whole galaxy | Quasars, blazars, variability |
| **Galaxy clusters** | Hundreds of galaxies in a shared dark-matter halo | Gravitational lensing, hot X-ray gas |
| **Cosmic web** | Filaments and voids — the largest structure there is | Redshift surveys |
| **The CMB** | Light released ~380,000 years after the Big Bang | Cosmology |

And separately, things that are **events rather than objects** — they appear, then fade:

| Transient | What happened | Timescale |
|---|---|---|
| **Flares** | Magnetic reconnection on a star's surface | Minutes–hours |
| **Novae** | Runaway fusion on a white dwarf's surface | Weeks |
| **Supernovae** | A massive star's core collapses, or a white dwarf detonates | Weeks–months |
| **Kilonovae** | Two neutron stars merge. Source of much of the gold in your phone | Days |
| **Tidal disruption events** | A star is pulled apart by a supermassive black hole | Months |
| **Gamma-ray bursts** | The most energetic explosions known | Seconds–minutes |
| **Fast radio bursts** | Millisecond radio flashes, mostly from magnetars | Milliseconds |

Transients are why *time-domain* astronomy — the thing you're learning — has become one of
the field's centres of gravity. You cannot find these in a single image.

## Scale, honestly

Distances are quoted in units chosen so the numbers stay manageable:

- **AU** — Earth–Sun distance, ~150 million km. For things inside a planetary system.
- **light-year** — how far light travels in a year, ~9.5 trillion km.
- **parsec (pc)** — 3.26 light-years. Defined by parallax geometry, so it's what
  astronomers actually use. kpc for galaxies, Mpc for the extragalactic universe.

Let's just plot it, because the ratios are the point.

```python
import matplotlib.pyplot as plt
import numpy as np

from skyplay import plotting

plotting.use_style()

# distance in metres, label
LADDER = [
    ('Earth radius',            6.37e6),
    ('Earth-Moon',              3.84e8),
    ('Sun radius',              6.96e8),
    ('Earth-Sun (1 AU)',        1.50e11),
    ('Neptune orbit (30 AU)',   4.50e12),
    ('Proxima Centauri (1.3 pc)', 4.01e16),
    ('Orion Nebula (400 pc)',   1.23e19),
    ('Galactic centre (8 kpc)', 2.47e20),
    ('Milky Way diameter',      9.5e20),
    ('Andromeda (0.78 Mpc)',    2.41e22),
    ('Virgo Cluster (16 Mpc)',  4.94e23),
    ('Observable universe',     4.4e26),
]

labels = [n for n, _ in LADDER]
metres = np.array([d for _, d in LADDER])

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(range(len(labels)), metres, color=plotting.SERIES[0], height=0.6)
ax.set_xscale('log')
ax.set_yticks(range(len(labels)), labels)
ax.set_xlabel('Distance (metres, log scale)')
ax.set_title('Every step here is a different universe')
ax.grid(True, axis='x', alpha=0.4)
plt.show()

print('Each tick is 10x. From your body to the observable universe is about 26 factors')
print('of ten -- which is why no mental picture survives the trip.')
```

## Distance is time

Light travels at a finite speed, so every observation is of the past. This is the single
most useful fact for keeping perspective on what "observing" means.

```python
from astropy import constants as const
from astropy import units as u

for name, dist in [('the Moon', 3.84e8 * u.m),
                   ('the Sun', 1 * u.AU),
                   ('Proxima Centauri', 1.3 * u.pc),
                   ('Kepler-8 (our target)', 1070 * u.pc),
                   ('the Galactic centre', 8 * u.kpc),
                   ('Andromeda', 0.78 * u.Mpc),
                   ('the most distant galaxies', 4 * u.Gpc)]:
    light_time = (dist / const.c).to(u.yr)
    if light_time < 1e-4 * u.yr:
        pretty = f'{light_time.to(u.s).value:8.2f} seconds'
    elif light_time < 1 * u.yr:
        pretty = f'{light_time.to(u.d).value:8.2f} days'
    else:
        pretty = f'{light_time.value:12,.0f} years'
    print(f'  {name:28s} {pretty} ago')

print()
print('The Kepler-8 light curves in this repo left the star before humans farmed.')
```

## What this means for the repo

Notice how little of that list is reachable with the technique you're learning. Transit
photometry sees **planets around nearby-ish stars**, full stop. That's not a weakness —
every technique has a narrow window, and progress comes from knowing exactly where yours
ends:

| You want to know about | Technique | Data product |
|---|---|---|
| Planets transiting a star | Time-series photometry | Light curve ← *you are here* |
| A planet's mass | Radial velocity (spectroscopy) | Spectra over time |
| A planet's atmosphere | Transmission spectroscopy | Spectra during transit |
| Stellar interiors | Asteroseismology | Light curve, frequency domain |
| Distances, motions | Astrometry (Gaia) | Positions over time |
| Galaxy composition, redshift | Spectroscopy | Spectrum |
| Dark matter distribution | Gravitational lensing | Images, or microlensing light curves |
| Merging compact objects | Gravitational waves | Strain time series |
